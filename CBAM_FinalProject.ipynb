```python
from ultralytics import YOLO

model = YOLO("ultralytics/cfg/models/ext/custom-yolo.yaml")

results = model.train(
    data="ultralytics/cfg/skyfusion/data.yaml",
    epochs=10,
    imgsz=640,
    batch=4,
    device=0,        # 0 = use the Colab GPU
    project="runs/final_project",
    name="cbam_doubleconv_exp",
    plots=True,
)
```

```python
from ultralytics import YOLO

model = YOLO("runs/detect/runs/final_project/cbam_doubleconv_exp/weights/best.pt")
metrics = model.val(data="ultralytics/cfg/skyfusion/data.yaml")

print(metrics.results_dict)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("precision:", metrics.box.mp)
print("recall:", metrics.box.mr)
```
